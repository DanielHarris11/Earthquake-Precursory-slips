```python
import os

# Function to rename files
def rename_files_in_directory(target_directory):
    # Walk through all subdirectories
    for root, dirs, files in os.walk(target_directory):
        for file in files:
            # Check if file starts with 'updated_' and ends with '.dat'
            if file.startswith("updated_") and file.endswith(".dat"):
                # Create new file name
                new_file_name = file.replace("updated_", "", 1)
                
                # Full old and new paths
                old_file_path = os.path.join(root, file)
                new_file_path = os.path.join(root, new_file_name)
                
                # Rename the file
                os.rename(old_file_path, new_file_path)
                print(f"Renamed: {old_file_path} -> {new_file_path}")

target_directory = r"../output"
rename_files_in_directory(target_directory)
```
